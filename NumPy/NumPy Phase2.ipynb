NumPy Array Operations

```python
import numpy as np
```

```python
arr=np.array([10,20,30,40,50,60,70])
print("Basic Slicing :",arr[:3])
print("Slicing with step:",arr[0:6:2])
print("-ve Indexing:",arr[-1])
```

```python
#sales year wise id 2023,2024,2025
sales=np.array([[0,10000,15000,18000],#Chocolate
                [1,14000,26000,32000],#Pizza
                [2,31000,24000,35000] #Ice cream
                ])
print("pizza sale 2024:",sales[1,2])
```
